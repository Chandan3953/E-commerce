```python
df = pd.read_csv("Sample - Superstore.csv",encoding = 'latin - 1')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

Coverting Date Column

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df['Order Month'] = df['Order Date'].dt.month
df['Order Year'] = df['Order Date'].dt.year
df['Order week of days'] = df['Order Date'].dt.dayofweek
```

```python
df.head()
```

Monthly Sales Analysis

```python
Monthly_sales = df.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
Monthly_sales.head()
```

```python
Monthly_sales = df.groupby('Order Month')['Sales'].sum().reset_index()
fig = px.line(Monthly_sales,
              x = 'Order Month',
              y = 'Sales',
              title = 'Monthly Sales Analysis')
fig.show()
```

Sales Analysis by Sub-Category

```python
sub_category = df.groupby('Sub-Category')['Sales'].sum().reset_index()
fig = px.line(sub_category,
              x = 'Sub-Category',
              y = 'Sales',
              title = 'SubCategory Sales Analysis')
fig.show()
```

Monthly Profit Analysis

```python
Monthly_profit = df.groupby('Order Month')['Profit'].sum().reset_index()
fig = px.line(Monthly_profit,
              x = 'Order Month',
              y = 'Profit',
              title = 'Monthly Profit  Sales Analysis')
fig.show()
```

Sales Analysis by Category

```python
sales_by_category = df.groupby('Category')['Sales'].sum().reset_index()
fig = px.pie(sales_by_category,
             values = 'Sales',
             names = 'Category',
             hole = 0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)
fig.update_traces(textposition = 'inside',textinfo = 'percent+label')
fig.update_layout(title = 'Sales Analysis by Category',title_font=dict(size=26))
fig.show()

    
```

Profit Analysis by Category

```python
Profit_by_category = df.groupby('Category')['Profit'].sum().reset_index()
fig = px.pie(Profit_by_category,
             values = 'Profit',
             names = 'Category',
             hole = 0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)
fig.update_traces(textposition = 'inside',textinfo = 'percent+label')
fig.update_layout(title = 'Profit Analysis by Category',title_font=dict(size=26))
fig.show()
```

Profit Analysis by Sub-Category

```python
Profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
fig = px.pie(Profit_by_subcategory,
             values = 'Profit',
             names = 'Sub-Category',
             hole = 0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)
fig.update_traces(textposition = 'inside',textinfo = 'percent+label')
fig.update_layout(title = 'Profit Analysis by SubCategory',title_font=dict(size=26))
fig.show()
```

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
fig = px.bar(profit_by_subcategory, x='Sub-Category', 
             y='Profit', 
             title='Profit Analysis by Sub-Category')
fig.show()
```

Sales and Profit Analysis by Customer Segment

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
